```python
import pandas as pd
from census import Census
from us import states

# ── Your Census API key ──
API_KEY = "YOUR_KEY_HERE"

c = Census(API_KEY)

# Sacramento region counties: Sacramento, Yolo, Placer, El Dorado
counties = {
    "Sacramento": "067",
    "Yolo": "113",
    "Placer": "061",
    "El Dorado": "017"
}

CA_FIPS = states.CA.fips

all_data = []

for county_name, county_fips in counties.items():
    print(f"Pulling data for {county_name} County...")
    
    result = c.acs5.state_county_tract(
        fields=(
            'NAME',
            'B17001_002E',  # Population below poverty level
            'B17001_001E',  # Total population (poverty universe)
            'B03002_003E',  # White alone, not Hispanic
            'B03002_004E',  # Black alone
            'B03002_012E',  # Hispanic or Latino
            'B25044_003E',  # Owner households, no vehicle
            'B25044_010E',  # Renter households, no vehicle
            'B25044_001E',  # Total households (vehicle universe)
        ),
        state_fips=CA_FIPS,
        county_fips=county_fips,
        tract="*",
        year=2022
    )
    
    df = pd.DataFrame(result)
    df['county'] = county_name
    all_data.append(df)

census_df = pd.concat(all_data, ignore_index=True)

census_df['poverty_rate'] = (
    pd.to_numeric(census_df['B17001_002E']) / 
    pd.to_numeric(census_df['B17001_001E']) * 100
).round(1)

census_df['no_vehicle_rate'] = (
    (pd.to_numeric(census_df['B25044_003E']) + pd.to_numeric(census_df['B25044_010E'])) /
    pd.to_numeric(census_df['B25044_001E']) * 100
).round(1)

census_df['pct_black'] = (
    pd.to_numeric(census_df['B03002_004E']) /
    pd.to_numeric(census_df['B17001_001E']) * 100
).round(1)

census_df['pct_hispanic'] = (
    pd.to_numeric(census_df['B03002_012E']) /
    pd.to_numeric(census_df['B17001_001E']) * 100
).round(1)

census_df.to_csv('sacramento_census_data.csv', index=False)

print(f"\nDone! {len(census_df)} tracts pulled across 4 counties.")
print(census_df[['NAME', 'county', 'poverty_rate', 'no_vehicle_rate', 'pct_black', 'pct_hispanic']].head(10))
```
